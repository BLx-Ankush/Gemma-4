# AEGIS: Gemma 4 Safety-Audited Medical First-Response Assistant

This notebook summarizes architecture, benchmark evidence, and demo narrative for the Gemma 4 Good hackathon submission.

## Architecture Highlights

- Online path: Gemma 4 cloud model for high-quality responses.
- Offline path: local Gemma 4 E2B fallback for connectivity resilience.
- Safety layer: MIRROR second-pass audit with verdict, confidence, and risk flags.
- Accessibility: voice input and text-to-speech output for every response.

```python
import json
from pathlib import Path

report_path = Path('benchmark/report.json')
if not report_path.exists():
    raise FileNotFoundError('benchmark/report.json not found. Run benchmark first.')

payload = json.loads(report_path.read_text(encoding='utf-8'))
report = payload['report']
results = payload['results']
payload['generated_at'], report['total']
```

```python
import pandas as pd

summary = pd.DataFrame([
    {'metric': 'Scenarios', 'value': report['total']},
    {'metric': 'Successful Requests', 'value': report['passed_requests']},
    {'metric': 'Latency P50 (ms)', 'value': report['latency']['p50_ms']},
    {'metric': 'Latency P95 (ms)', 'value': report['latency']['p95_ms']},
    {'metric': 'Safety Caught', 'value': report['safety']['caught_warn_or_block']},
    {'metric': 'Dangerous Scenarios', 'value': report['safety']['dangerous_scenarios']},
])
summary
```

```python
import matplotlib.pyplot as plt

verdicts = report['mirror_verdicts']
labels = ['pass', 'warn', 'block']
values = [verdicts.get(label, 0) for label in labels]
colors = ['#0db89e', '#e07b20', '#c6432d']

plt.figure(figsize=(6, 3.5))
plt.bar(labels, values, color=colors)
plt.title('MIRROR Verdict Distribution')
plt.ylabel('Count')
plt.tight_layout()
plt.show()
```

## Responsible Use and Limits

AEGIS is for first-response support and triage guidance. It does **not** replace professional diagnosis, emergency care, or licensed medical judgment.
